```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
file_path = r"C:\Users\ebruk\OneDrive\Desktop\charity_donation_data_final.csv" 
df = pd.read_csv(file_path)

# Display the first few rows
print("Dataset Preview:")
display(df.head())

# Basic exploration
print("\nDataset Info:")
print(df.info())

print("\nSummary Statistics:")
print(df.describe())

# Check for missing values
print("\nMissing Values Per Column:")
print(df.isnull().sum())

```

```python
# Convert Donation_Date to datetime
df['Donation_Date'] = pd.to_datetime(df['Donation_Date'], errors='coerce')

# Check for invalid dates (if any)
invalid_dates = df[df['Donation_Date'].isnull()]
if not invalid_dates.empty:
    print("Invalid Donation Dates Found:")
    print(invalid_dates)

# Extract Year and Month from Donation_Date
df['Donation_Year'] = df['Donation_Date'].dt.year
df['Donation_Month'] = df['Donation_Date'].dt.month

# Check for outliers in Donation_Amount
print("\nChecking for outliers in Donation_Amount:")
print(df['Donation_Amount'].describe())

# Optional: Remove or cap outliers (if necessary)
df['Donation_Amount'] = np.where(
    df['Donation_Amount'] < 0, 0, df['Donation_Amount']  # Replace negative donations with 0
)

# Create a grouping to check seasonal donation patterns
seasonal_donations = df.groupby('Donation_Season')['Donation_Amount'].sum()

print("\nSeasonal Donation Patterns:")
print(seasonal_donations)
```
